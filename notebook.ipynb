# OilyGiant Oil Well Development Project

## Introduction

This project was developed for **OilyGiant**, an oil extraction company seeking to identify the most profitable region for developing **200 new oil wells**.

Using geological exploration data from three different regions, a **Linear Regression** model was built to predict the volume of oil reserves in potential wells. The predictions were then used to estimate potential profits under specific business constraints, including development costs and oil revenue.

To support decision-making, a **Bootstrapping** analysis was performed to evaluate expected profits, confidence intervals, and the probability of financial losses. The final recommendation is based on both profitability and risk, ensuring that only regions with a loss probability below **2.5%** are considered.

## Data Loading and Exploration

```python
import pandas as pd
import os
```

```python
# Load the datasets
data = []

for i in range(0, 3):
    file_path = f'datasets/geo_data_{i}.csv' # Local path
    if not os.path.exists(file_path):
        file_path = '/'+file_path # Platform path
    df = pd.read_csv(file_path)
    data.append(df)
```

```python
# Check the datasets
for df in data:
    display(df.head())
    print(df.info())
    print(f'Duplicates: {df.duplicated().sum()}')
    print('-'*80)
```

The datasets do not contain any missing values or duplicate records.

## Data Preparation

```python
# Prepare the datasets for the model
for df in data:
    # Drop Id column
    df.drop(columns=['id'], inplace=True)
```

## Model Training and Evaluation

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Function that divides the datasets into features and target, trains the model and analyzes the results
def train_model(data):
    # Split the datasets into features and target
    x = data.drop(columns=['product'])
    y = data['product']
    
    # Train-test split
    train_x, test_x, train_y, test_y = train_test_split(x, y, test_size=0.25, random_state=12345)
    
    # Model training
    model = LinearRegression()
    model.fit(train_x, train_y)
    
    # Predictions
    predictions = model.predict(test_x)
    
    # Metrics
    rmse = mean_squared_error(test_y, predictions)**0.5
    r2 = r2_score(test_y, predictions)
    mean_volume = predictions.mean()
    
    return {'predictions': predictions, 'test_y': test_y, 'rmse': rmse, 'r2': r2, 'mean_volume': mean_volume}
```

```python
metrics_list = []

for data_set in data:
    metrics_dict = train_model(data_set)
    print(f"Metrics for dataset:\nR2: {metrics_dict['r2']:.2f}\nRMSE: {metrics_dict['rmse']:.2f}\nMean Volume: {metrics_dict['mean_volume']:.2f}")
    print('-'*80)
    metrics_list.append(metrics_dict)
```
